IncrementalXGBoost es una clase wraper que se encarga de ejecutar de forma invariante el modelo. 
Funciona asi:
  1. Estamos en la semana t, se predice para la semana t+1, ve a 2.
  2. Existen labels para la semana t+1? : Si hay, ve a 3. Si no, ve a 6.
  3. Hay labels, evalúa el desempeño de la predicción t+1. Si (eval(t) - eval(t+1) < umbral), ve a 4, si no, ve a 5  
  4. Hacer un retrain con el historial de semanas, ve a 1 pero con t+2.
  6. Hacer un incremental_training sobre el modelo de XGboost. Ve a 1 con t+2.
  5. Volver a 1. pero con semana t+2
  6. Terminar y esperar a nuevas labels.

Para obtener las predicciones utilizaremos modelos entrenados hasta la semana t y predeciremos la semana t+1.

### 📌 Análisis comparativo [2.0 puntos]

````python
def metrics(labels,  predictions, batch : str): 
    f1 = f1_score(labels, predictions, average='binary')  # Ajusta el average si es multiclase
    print(f"F1 Score: {f1:.4f}")
    
    cm = confusion_matrix(labels, predictions)
    tn, fp, fn, tp = confusion_matrix(labels, predictions).ravel()
    false_positive_rate = fp / (fp + tn) if (fp + tn) > 0 else 0
    false_negative_rate = fn / (fn + tp) if (fn + tp) > 0 else 0
    precision = tp / (tp + fp) if (tp + fp) > 0 else 0
    recall = tp / (tp + fn) if (tp + fn) > 0 else 0
    specificity = tn / (tn + fp) if (tn + fp) > 0 else 0
    print(f' False positive rate {false_positive_rate}\n False negative rate {false_negative_rate} \n Precision {precision} \n Recall {recall} \n Specificity {specificity}' )
    sns.set_theme(style="darkgrid")
    plt.figure(figsize=(6, 4))
    colors = sns.color_palette("deep", as_cmap=True)
    sns.heatmap(cm, annot=True, fmt='d', cmap = colors, cbar=False,
                xticklabels=["Pred. 0", "Pred. 1"],
                yticklabels=["True 0", "True 1"])
    plt.xlabel("Predicted")
    plt.ylabel("True Label")
    plt.title(f"Confusion Matrix, batch : {batch}")
    plt.tight_layout()
    plt.show()
````
